here we will start to build the windows

```python
import rasterio
import numpy as np
import pandas as pd
import os 
df = pd.read_excel("results/soil_dataset.xlsx", sheet_name="scorpan")
df = df.dropna(subset=["X", "Y"])

def latlon_to_pixel(lat, lon, transform):
    if np.isnan(lat) or np.isnan(lon):
        return None
    
    pixel_size_x = transform[0]  
    pixel_size_y = transform[4]   
    origin_x = transform[2]       
    origin_y = transform[5]       

    col = (lon - origin_x) / pixel_size_x
    row = (origin_y - lat) / -pixel_size_y  # Invert pixel size for latitude

    return int(round(row)), int(round(col))
# used to save the ordre of the covariables
covariables=[]
# used to store all the windows ta3 chaque X , Y Nx7x7x6
df_patches = []
nb_covariables = 6
for i in os.listdir("Maps"):
    raster_path = f"Maps/{i}/wc2.1_30s_{i}_01.tif"
    covariables.append(i)  # to keep the order
    with rasterio.open(raster_path) as src:
        raster_data = src.read()
        transform = src.transform  # transformation matrix from lat lon to pixels
        print(src.crs)
        print(transform)

    patch_size = (7, 7 , nb_covariables)
    half_h, half_w = patch_size[0] // 2, patch_size[1] // 2
    
    for idx, row in df.iterrows():
        lat, lon = row["X"], row["Y"]
        # transformation to pixels 
        pixel_y, pixel_x = latlon_to_pixel(lat, lon, transform)
        top, bottom = max(0, pixel_y - half_h), min(raster_data.shape[1], pixel_y + half_h + 1)
        left, right = max(0, pixel_x - half_w), min(raster_data.shape[2], pixel_x + half_w + 1)
        if len(covariables)==1:
            window = np.zeros((patch_size[0], patch_size[1], patch_size[2]))
            df_patches.append(window)
        
        df_patches[idx][:,:,len(covariables)-1] = raster_data[:, top:bottom, left:right]
    
df_patches = np.array(df_patches)
print(df_patches.shape)

```

each one represent the window for a specific covariant
