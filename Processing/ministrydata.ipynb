```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv("../results/ministere.csv")
df
```

```python
rename_map = {
    "X": "X",
    "Y": "Y",
    "Z": "altitude",
    "Depth": "Depth",
    "pH": "pH",
    "Conductivité (mmhos/cm)": "CE",
    "Cation Exchange Capacity" if "CEC" not in df.columns else "CEC": "CEC",  # Safeguard
    "Argile (%)": "A%",
    "Limon fin (%)": "L%",
    "Sable fin (%)": "S%",
    "Fe2O3 libre %": "ALR(L)",
    "Fe2O3 total %": "ALR(S)",
    "Matière organique (%)": "OM"
}

df = df.rename(columns=rename_map)
```

```python
df
```

```python
df.columns
```

```python
df[["UD", "LD"]] = df["Depth"].str.extract(r'(\d+)[^\d]+(\d+)').astype(float)
df["average profondeur"] = (df["UD"] + df["LD"]) / 2
df["USDA Soil Taxonomy"] = "To classify"
keep_cols = [
    "X", "Y", "altitude", "UD", "LD", "average profondeur",
    "USDA Soil Taxonomy", "pH", "CE", 
    "A%", "L%", "S%", "ALR(L)", "ALR(S)", "OM"
]
df_final = df[keep_cols]
df_final
```

```python
df_final["CEC"]=np.nan
df_final
```

```python
df_final.info()
```

```python
from pyproj import Transformer

# Lambert Nord Sahara (EPSG:30491) -> WGS84 (EPSG:4326)
transformer = Transformer.from_crs("EPSG:30491", "EPSG:4326", always_xy=True)

for i in range(len(df)):
    x = df_final.at[i, "X"] * 1000
    y = df_final.at[i, "Y"] * 1000
    lon, lat = transformer.transform(x, y)
    df_final.at[i, "X"] = round(lon, 2) 
    df_final.at[i, "Y"] = round(lat, 2)  
```

```python
df_final
```

```python
df_final.to_csv("../results/myministereData.csv")
```
