```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler
```

1. get the cleaned ministery data

```python
df_ministere = pd.read_csv("../../results/Soil/myministereData.csv")
df_ministere = df_ministere.drop(columns=["Unnamed: 0"])
df_ministere.info()
```

2. get the cleaned PDF and given data Combined

```python
df_pdf = pd.read_csv("../../results/Soil/CombinedData.csv")
df_pdf = df_pdf.drop(columns=["Unnamed: 0.1" , "Unnamed: 0"])
df_pdf.info()
```

```python
df = pd.concat([df_pdf , df_ministere])
df = df.drop(columns=["altitude" , "USDA Soil Taxonomy"])
df = df.dropna(subset=["X" , "Y"])
df
```

3. Treating the missing values

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_df)

imputer = KNNImputer(n_neighbors=5)
imputed_array = imputer.fit_transform(scaled_data)
real_values = scaler.inverse_transform(imputed_array)
df[numeric_df.columns] = real_values
df.info()
```

4. Saving the dataset

```python
df.to_csv("../../results/Soil/final_Soil_dataset.csv")
```
