```python
import pandas as pd
import numpy as np

df = pd.read_csv("full_reviewed.csv")
df = df[["X","Y","Debut Profondeur" ,"Fin Profondeur" , "ALR(S)"]]
df = df.dropna(subset=["X","Y","Debut Profondeur" ,"Fin Profondeur"])
for i in df.columns:
    print(i)
    if i in ["Debut Profondeur", "Fin Profondeur"]:
        df[i] = df[i].apply(lambda x: int(float(str(x).replace(",", "."))) if pd.notna(x) else np.nan)
        df[i]= df[i].astype("int")
    else:
        df[i] = df[i].apply(lambda x: float(str(x).replace(",", ".")) if pd.notna(x) else np.nan)

def format_sid(x, y):
    return f"{x}_{y}"

df['SID'] = df.apply(lambda row: format_sid(row['X'], row['Y']), axis=1)
df = df[['SID', 'X', 'Y', 'Debut Profondeur', 'Fin Profondeur', 'ALR(S)']]
df
```

```python
%%R

df <- read.csv("test.csv")
# Rename columns
str(df)
names(df)[names(df) == "Debut.Profondeur"] <- "UD"
names(df)[names(df) == "Fin.Profondeur"] <- "LD"
names(df)[names(df) == "ALR.S."] <- "VAL"

df$VAL <- as.numeric(gsub(",", ".", df$VAL))
df$X <- as.numeric(gsub(",", ".", df$X))
df$Y <- as.numeric(gsub(",", ".", df$Y))

str(df)
df <- df[!is.na(df$UD) & !is.na(df$LD) & df$UD < df$LD, ]

valid_SIDs <- df %>%
  group_by(SID) %>%
  filter(n() >= 2) %>%
  pull(SID) %>%
  unique() %>%
  as.character()

df <- df[df$SID %in% valid_SIDs, ]
```

```python
%%R
df <- df[, !(names(df) %in% c("X", "Y" , "X.1"))]
str(df)
spl_dat <- mpspline_tidy(obj = df, var_name = 'VAL', )
ph_spline_df <- spl_dat$est_dcm
write.csv(ph_spline_df, "ALR(S)_spline_result.csv", row.names = FALSE)
```
