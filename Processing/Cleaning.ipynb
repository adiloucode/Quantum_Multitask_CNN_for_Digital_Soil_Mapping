3. function transforme the lat and lon into pixels

```python
def latlon_to_pixel(lat, lon, transform):
    if np.isnan(lat) or np.isnan(lon):
        return None
    
    pixel_size_x = transform[0]  
    pixel_size_y = transform[4]   
    origin_x = transform[2]       
    origin_y = transform[5]       

    col = (lon - origin_x) / pixel_size_x
    row = (origin_y - lat) / -pixel_size_y  # Invert pixel size for latitude

    return int(round(row)), int(round(col))
```

4. windows building for basic CNN model (inputs outputs)

```python
def build_data(df , length , width , nb_covariables , column_name) : 
    # used to save the ordre of the covariables
    covariables=[]
    # used to store all the windows ta3 chaque X , Y Nx7x7x6
    df_patches = []
    final_outputs = []
    for index , i in enumerate(os.listdir("Maps")):
        covariables.append(i)  # to keep the order
        print(f"file : Maps/{i}/wc2.1_30s_{i}_01.tif")
        raster_path = f"Maps/{i}/wc2.1_30s_{i}_01.tif"
        with rasterio.open(raster_path) as src:
            raster_data = src.read()
            transform = src.transform  # transformation matrix from lat lon to pixels

        patch_size = (length, width , nb_covariables)
        half_h, half_w = patch_size[0] // 2, patch_size[1] // 2
        for idx, row in df.iterrows():
            lat, lon = row["X"], row["Y"]
            # transformation to pixels 
            pixel_y, pixel_x = latlon_to_pixel(lat, lon, transform)
            top, bottom = max(0, pixel_y - half_h), min(raster_data.shape[1], pixel_y + half_h + 1)
            left, right = max(0, pixel_x - half_w), min(raster_data.shape[2], pixel_x + half_w + 1)
            
            if len(covariables)==1:
                window = np.zeros((patch_size[0], patch_size[1], patch_size[2]))
                df_patches.append(window)
                
            df_patches[idx][:,:,len(covariables)-1] = raster_data[:, top:bottom, left:right]    
            output = []

            if index ==0:
                output.append(row[column_name])
                output = np.array(output)
                output = output.reshape((len(output),1))
                final_outputs.append(output)
        print(f"length of df patches is : {len(df_patches)} and final outputs is {len(final_outputs)}")
    df_patches = np.array(df_patches)
    print(df_patches.shape)
    return covariables,df_patches ,final_outputs # liste ta3 covariantes , inputs , outputs  

covariables , df_patches , final_outputs = build_data(df , 7 , 7 , 6 , "pH")
```

2. forming the outputs based on a range of depths

```python
def outputs_building(atr):
    depth_ranges = [
        (0, 15),
        (5, 15),
        (15, 30),
        (30, 60),
        (60, 100),
        (100, 200)
    ]
    result_array = np.zeros((496, 6))
    depth_indices = {i: [] for i in range(len(depth_ranges))}
    for i, (lb, ub) in enumerate(depth_ranges):
        for j in range(lb, ub + 1):
            depth_indices[i].append(j)
    for _, row in df.iterrows():
        x, y, ud, ld, val = row['X'], row['Y'], row['UD'], row['LD'] , row[atr]
        for index, depth_range in depth_indices.items():
            if any(d in range(int(ud), int(ld) + 1) for d in depth_range):
                result_array[df_unique[(df_unique['X'] == x) & (df_unique['Y'] == y)].index[0], index] = val

    mean = df[atr].mean()
    result_array = np.nan_to_num(result_array, nan=mean)
    finals = np.array(result_array).reshape(496,6,1)
    for i in finals:
        print(i)
    
    print(finals.shape)
    return finals
```

```python
for i in df.columns[6:]:
    print(i)
    finals = outputs_building(i)
    np.save(f'../Data_multi_CNN/final_data/output_v6/output{i}.npy', np.array(finals))
```
