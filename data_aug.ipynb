```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
inputs = np.load("Data_multi_CNN/inputs.npy")
outputs = np.load("Data_multi_CNN/output.npy")
inputs.shape
```

```python
def augment_data(data, output):
    augmented_data = []
    augmented_output = []

    for i in range(data.shape[0]):
        original = data[i]
        original_output = output[i]
        flipped_horizontal = np.flip(original, axis=1)  #horizontally
        flipped_vertical = np.flip(original, axis=0)    #vertically

        rotated_90 = np.rot90(original, k=1)  #90°
        rotated_180 = np.rot90(original, k=2)  #180°
        rotated_270 = np.rot90(original, k=3)  #270°

        augmented_data.extend([
            original,
            flipped_horizontal,
            flipped_vertical,
            rotated_90,
            rotated_180,
            rotated_270
        ])
        # outputs
        augmented_output.extend([
            original_output,
            original_output,
            original_output,
            original_output,
            original_output,
            original_output
        ])

    return np.array(augmented_data), np.array(augmented_output)

augmented_data, augmented_output = augment_data(inputs, outputs)
print("original data shape:", inputs.shape)
print("augmented data shape:", augmented_data.shape)
print("original output shape:", outputs.shape)
print("augmented output shape:", augmented_output.shape)
```

```python
np.save('Data_multi_CNN/input_aug.npy', augmented_data)
np.save('Data_multi_CNN/output_aug.npy', np.array(augmented_output))
```
