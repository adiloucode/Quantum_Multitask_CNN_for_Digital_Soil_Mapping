```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
import numpy as np
from sklearn.model_selection import train_test_split

def build_multi_task_cnn(input_shape=(7, 7, 6)):
    input_layer = keras.Input(shape=input_shape)

    # Convolutional Layers with Batch Normalization
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(input_layer)
    x = layers.BatchNormalization()(x)  # Adding Batch Normalization
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Flatten()(x)

    outputs = []
    for _ in range(6):
        branch = layers.Dense(32, activation='relu')(x)  # Increased units
        branch = layers.Dropout(0.3)(branch)
        branch = layers.Dense(16, activation='relu')(branch)  # Reduced units
        branch = layers.Dense(1, activation='linear')(branch)
        outputs.append(branch)

    model = keras.Model(inputs=input_layer, outputs=outputs)

    # Compile with learning rate scheduling
    optimizer = keras.optimizers.Adam(learning_rate=0.001)
    model.compile(optimizer=optimizer, loss='mse', metrics=['mae' , 'mae','mae' , 'mae','mae' , 'mae'])

    return model

# Example setup
input_shape = (7, 7, 6)
model = build_multi_task_cnn(input_shape)
model.summary()
```

```python
inputs = np.load('input_aug.npy' , allow_pickle=True)
outputs = np.load('output_aug.npy')
x_train , x_test , y_train , y_test = train_test_split(inputs , outputs , test_size=0.1)
print(inputs.shape)
print(outputs.shape)
print(len(x_train))
print(len(x_test))
print(len(y_train))
print(len(y_test))

y_train_split = [y_train[:, i, 0] for i in range(y_train.shape[1])]
y_test_split = [y_test[:, i, 0] for i in range(y_test.shape[1])]
model.fit(x_train, y_train_split, epochs=100, batch_size=8, validation_data=(x_test, y_test_split))
```
