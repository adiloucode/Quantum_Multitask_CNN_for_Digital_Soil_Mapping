```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
import numpy as np
from sklearn.model_selection import train_test_split

def build_multi_task_cnn(input_shape=(None, None, None)):
    input_layer = keras.Input(shape=input_shape)
    x = layers.Conv2D(16, (3, 3), activation='relu', padding='same')(input_layer)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    x = layers.Flatten()(x)
    outputs = []

    for _ in range(6):
        branch = layers.Dense(10, activation='relu')(x)
        branch = layers.Dropout(0.3)(branch)
        branch = layers.Dense(10, activation='relu')(branch)
        branch = layers.Dense(1, activation='linear')(branch)
        outputs.append(branch)

    model = keras.Model(inputs=input_layer, outputs=outputs)
    model.compile(optimizer='adam', loss='mse', metrics=['mae' , 'mae','mae' , 'mae','mae' , 'mae'])

    return model

input_shape = (7, 7, 6)
model = build_multi_task_cnn(input_shape)
model.summary()
```

```python
inputs = np.load('inputs.npy')
outputs = np.load('output.npy')
x_train , x_test , y_train , y_test = train_test_split(inputs , outputs , test_size=0.1)
```
